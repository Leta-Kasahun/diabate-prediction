# Data Cleaning

## Findings
- The original dataset contains **403 records and 19 columns**, representing the available observations and variables.
- The columns `bp.2s` and `bp.2d` have approximately **65% missing values**, meaning most records do not contain measurements for these variables.

## Cleaning Decisions
- Removed `bp.2s` and `bp.2d` because their high percentage of missing values could reduce the reliability of these features.
- Removed unnecessary leading and trailing whitespace from categorical values to ensure consistent category labels during later encoding.
- Retained `id` for record identification and tracking, but excluded it from the model features because it does not represent a meaningful predictive measurement.
- Retained other missing values for handling during the preprocessing stage, where appropriate imputation methods will be applied.

## Result
- The cleaned dataset contains **403 records and 17 columns**.
- The original dataset remains unchanged, and the cleaned dataset is saved for the subsequent data preparation stages.
