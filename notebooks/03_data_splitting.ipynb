# Data Splitting

## Objective
The purpose of data splitting is to divide the dataset into training and testing sets. The training set is used to train machine-learning models, while the testing set is reserved for evaluating their performance on unseen data.

## Data Preparation
- Selected `glyhb` as the target variable because the assignment focuses on predicting GlyHb percentages.
- Removed records with missing `glyhb` values because supervised regression requires known target values during training.
- Excluded `id` from the input features because it is used only for record identification and does not represent a meaningful predictive measurement.
- Separated the dataset into:
  - **Features (X):** 15 input variables used for prediction.
  - **Target (y):** GlyHb values to be predicted.

## Train-Test Split
- Applied an **80:20 split**:
  - 80% for training.
  - 20% for testing.
- Used `random_state=42` to ensure reproducibility and maintain the same data partition across model experiments.
- Used the same training and testing sets for comparing regression algorithms to ensure a fair evaluation.

## Results
- Records available after removing missing target values: **390**.
- Training set: **312 records**.
- Testing set: **78 records**.
- Each saved dataset contains 15 input features and the `glyhb` target column, giving 16 columns in total.

## Saving the Datasets
The training and testing datasets were saved in `data/processed/` as:
- `train.csv`
- `test.csv`

These saved datasets allow subsequent experiments to use the same data partitions consistently.
